```python
import pandas as pd

# Einlesen der CSV-Datei mit den Schichtentemperaturen
df_schichten = pd.read_csv('Outputs/df_schichten.csv', delimiter=',', header=0)

ax = df_schichten.plot()
ax.set_xlim(left=0)
ax.set_ylabel('Temperatur')
ax.set_xlabel('Zeit')
ax.figure.savefig('Outputs/python2plot.png')
```

```python
# Einlesen der CSV-Datei mit den Schichtentemperaturen aus TRNSYS
#Gradzahlen einheitlich machen
import pandas as pd
 
df = pd.read_csv('TRNSYSspeicherschichttemperaturen.csv', skiprows=[1], encoding='latin1')
df = df.dropna(axis=1, how='all')
df.columns = df.columns.str.strip()
df.drop(columns=['TIME'], inplace=True)

ax = df.plot()
ax.set_xlim(left=0)
ax.set_ylabel('Temperatur')
ax.set_xlabel('Zeit')
ax.figure.savefig('Outputs/trynsis2plot.png')
```
